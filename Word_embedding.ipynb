```python
sentences =[
"my country is india and it got independence on the month of august.",
"my country has more than 50 cities.",
"my country has a total population of 1.1 billion when the last census were done.",

"the technology that i am most interested is AR." ,
"the reason i like AR is that using it we can stay in relity at the same time in the virtual world.",
"using AR i can watch how to make food and cook while watching the steps at the same time.",

"my favourite sport is playing and watching football.",
"the team that i support the most is brazil because that was the first team i got to know and not  other teams.",

"the the the india india india technology, technology technology, AR Ar AR"
]
```

```python
# imprort the libraries
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer #TF-bow
from sklearn.feature_extraction.text import TfidfVectorizer #TFIFD
from sklearn.metrics.pairwise import cosine_similarity
```

```python
#embedding using TF-bow approach
tf_vectorizer=CountVectorizer()
tf_matrix=tf_vectorizer.fit_transform(sentences)
```

```python
tf_matrix
```

```python
tf_similarity=cosine_similarity(tf_matrix)
tf_similarity
```

```python
#embedding using TFidf approach
tfidf_vectorizer=TfidfVectorizer()
tfidf_matrix=tfidf_vectorizer.fit_transform(sentences)
```

```python
tfidf_matrix
```

```python
tfidf_similarity=cosine_similarity(tfidf_matrix)
tfidf_similarity
```

```python
#finding out the document similarity
def docsim(similarity_matrix, sentences):
  for i in range(len(sentences)):
    sims=similarity_matrix[i].copy()
    sims[i]=-1
    top_index=np.argmax(sims)
    print(sentences[i],"-->", sentences[top_index])
```

```python
docsim(tfidf_similarity,sentences)
```
